## **Handling Duplicate**
Data duplikat ditemukan berdasarkan kolom `Name`, `Platform`, dan `Year`

```python
# Menampilkan duplikat dari kolom nama, platform dan year
df[df.duplicated(subset=['Name', 'Platform', 'Year'], keep=False)]
```

```python
# Menghapus baris duplikat Global sales terkecil
df = df.drop_duplicates(subset=['Name', 'Platform', 'Year'], keep='last')

# Buktikan duplikat sudah terhapus
df[df.duplicated(subset=['Name', 'Platform', 'Year'], keep=False)]
```

## **Handling Missing Values**

Missing values ditemukan pada kolom `Publisher` terdapat `NaN`, 'Number None' dan 'Unknown' maka diseragamkan dengan 'Unknown Publisher'

```python
# Mengisi kolom Publisher
# Ubah nilai 'NaN', 'Unknown', dan 'Number None' menjadi 'Unknown Publisher'
df["Publisher"] = df["Publisher"].fillna("Unknown Publisher")
df.loc[
    df["Publisher"].isin(["Unknown", "Number None"]), "Publisher"
] = "Unknown Publisher"
```

```python
# Mengisi null pada kolom city dan country dengan 'Not Specified'
df["City"] = df["City"].fillna("Not Specified")
df["Country"] = df["Country"].fillna("Not Specified")
```

```python
# Cek tabel setelah missing values
df[['Name', 'City', 'Country', 'Publisher']]
```

```python
# Cek tabel setelah handling missing values
df.isna().sum()
```

## **Handling Tipe Data dan Penyelarasan**

```python
# Mengubah tipe data kolom Year ke==dari float64 menjadi numeric atau INT
df["Year"] = pd.to_numeric(df["Year"], errors="coerce").astype(int)
```

```python
# Cek ulang tipe data
df.dtypes
```

```python
# Pencocokan pembulatan Global_Sales dengan total penjumlahan regional sales (NA + EU + JP + Other)
def perbaiki_global_sales (row):
  # Hitung ulang total penjualan regional
  calc = row["NA_Sales"] + row["EU_Sales"] + row["JP_Sales"] + row["Other_Sales"]
  diff = abs(row["Global_Sales"] - calc)

  # Menggunakan logika if-else
  if 0.001 < diff <= 0.05:
    return calc
  else:
    return row["Global_Sales"]


# Apply fungsi ke seluruh baris DataFrame
df["Global_Sales"] = df.apply(perbaiki_global_sales, axis=1)
```

```python
# Cek Global Sales
df[['Name', 'NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales', 'Global_Sales']]
```

## **Handling Inkonsistensi**
Inkonsistensi ditemukan pada kolom `Country` yang mana game milik publisher 'Ubisoft' diisi dengan nama benua yaitu 'Europe' sedangkan yang lainnya diisi dengan nama negara yaitu 'France'

```python
# Ubah nama benua 'Europe' menjadi Country 'France
df.loc[
    (df["Publisher"] == "Ubisoft") & (df["Country"] == "Europe"), "Country"
] = "France"
```

```python
df[['Name', 'City', 'Country', 'Publisher']]
```

```python
# 1. Hitung jumlah platform unik per nama game
platform_counts = (
    df.groupby("Name")["Platform"].nunique().reset_index(name="num_platforms")
)

# 2. Gabungkan kembali ke dataframe utama
df = pd.merge(df, platform_counts, on="Name")

# 3. Buat kolom dimensi Release Strategy dengan benar
df["Release Strategy"] = df["num_platforms"].apply(
    lambda x: "Exclusive (1 Platform)"
    if x == 1
    else "Multi-Platform (2+ Platforms)"
)
```

```python
# Simpan data yang sudah di cleaning

df.to_excel('VG_Sales_Cleaned.xlsx', index=False)
```
